```python
from logging import root

from data_preprocessing import downsample_audio,split_audio

import os
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent.parent
root_dir = PROJECT_ROOT
data_dir = root_dir / "data"
violin = data_dir / "raw" / "solo_violin"
processed_dir = data_dir / "preprocessed" / "solo_violin"
processed_dir.mkdir(parents=True, exist_ok=True)
```

```python
from data_preprocessing import clip_silence
from utils import load_audio
from visualize import plot_spectrograms

# Demo: trim silence from one raw file and compare spectrograms before/after.
demo_file = next((violin / f for f in os.listdir(violin) if f.endswith('.wav')), None)
if demo_file is None:
    raise FileNotFoundError(f'No .wav files found in {violin}')

demo_out_dir = data_dir / 'preprocessed' / 'clip_demo'
demo_out_dir.mkdir(parents=True, exist_ok=True)
demo_out_path = demo_out_dir / f'{demo_file.stem}_trimmed.wav'

before_audio, before_sr = load_audio(demo_file, sr=None, mono=True)
after_audio, after_sr, (start_sample, end_sample) = clip_silence(
    demo_file,
    out_path=demo_out_path,
    threshold_db=-40.0,
)

start_sec = start_sample / after_sr
end_sec = end_sample / after_sr
print(f'Input file: {demo_file}')
print(f'Trimmed file saved to: {demo_out_path}')
print(f'Kept segment: samples [{start_sample}:{end_sample}] (~{start_sec:.2f}s to ~{end_sec:.2f}s)')

plot_spectrograms([before_audio, after_audio], fs=after_sr)
```
