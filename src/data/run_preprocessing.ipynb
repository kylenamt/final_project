```python
from logging import root

from preprocessing import downsample_audio,split_audio

import os
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent.parent
root_dir = PROJECT_ROOT
data_dir = root_dir / "data"
violin = data_dir / "raw" / "solo_violin"
processed_dir = data_dir / "preprocessed" / "solo_violin"
processed_dir.mkdir(parents=True, exist_ok=True)

for file in os.listdir(violin):
    if file.endswith(".wav"):
        in_path = violin / file
        out_path = processed_dir / file
        out_path, sr = downsample_audio(in_path, out_path, target_sr=16000)
        split_audio(out_path, processed_dir, n_seconds=4)
        os.remove(out_path)
```

```python
num_file = 0
for i in os.listdir(processed_dir):
    if i.endswith(".wav"):
        num_file += 1
print(f"Number of processed files: {num_file}")

    
```
